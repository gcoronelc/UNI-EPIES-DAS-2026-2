# Trabajando con PandaSerie
Para importar pandas en su script o notebook, la convención estándar de la
comunidad es usar el alias pd (pandas Development Team, 2024):

```python
# Importando pandas
# Se recomienda también importar NumPy
import pandas as pd
import numpy as np 

# Verificar la versión instalada
print(pd.__version__) # Ejemplo de salida: 3.1.0
```

# Eemplo 1
Crear una Series desde una lista

```python
notas = pd.Series([15, 18, 12, 20, 14])
print(notas)
```

# Eemplo 2
# Desde una lista con índice personalizado
Es posible asignar etiquetas descriptivas al índice para hacer más legible el acceso a
los datos:

```python
alumnos = ['Ana', 'Luis', 'María', 'Carlos', 'Elena']
notas = [15, 18, 12, 20, 14]
alu_notas = pd.Series(notas, index=alumnos, name='Notas_Finales')
print(alu_notas)

print(f"Nota de Ana:{alu_notas['Carlos']}")
```

# Ejemplo 3
# Desde un diccionario de Python
Cuando se crea una Series desde un diccionario, las claves se convierten en el índice
y los valores se convierten en los datos (VanderPlas, 2023):

```python
poblacion = {
    'Lima': 10900000,
    'Arequipa': 1080000,
    'Trujillo': 990000,
    'Chiclayo': 600000
}
serie_poblacion = pd.Series(poblacion, name='Población_2026')
print(serie_poblacion)

ciudad = "Chiclayo"
print(f"Poblacion de {ciudad}: {serie_poblacion[ciudad]}")
```

```python
print(serie_poblacion.index)
```

```python
print(serie_poblacion.values)
```

# Ejemplo 4
Propiedades de la serie panda

```python
s = pd.Series([10, 20, 30, 40, 50],
    index=['a', 'b', 'c', 'd', 'e'],
    name='ejemplo')

print('Valores: ', s.values) # ndarray de NumPy
print('Índice: ', s.index) # RangeIndex o Index
print('Tipo: ', s.dtype) # tipo de dato
print('Tamaño: ', s.size) # número de elementos
print('Forma: ', s.shape) # tupla (n,)
print('Nombre: ', s.name) # nombre de la Serie
```

# Ejemplo 5
# Acceso por etiqueta con .loc[]
.loc[] permite seleccionar elementos por su etiqueta de índice. Es el método
recomendado cuando se trabaja con índices con nombres:

```python
precios = pd.Series({
    'manzana': 2.5, 
    'pera': 3.0,
    'uva': 5.5, 
    'mango': 4.0,
    'papaya':6.0,
    'fresa':7.5
})

print("Serie creada")
```

```python
# Acceso a un solo elemento
print(precios.loc['pera']) # 3.0
print("-" * 20)

# Acceso a múltiples elementos (lista de etiquetas)
print(precios.loc[['pera', 'papaya']])
print("-" * 20)

# Slicing por etiqueta (incluye extremo final)
print(precios.loc['pera':'papaya'])
```
